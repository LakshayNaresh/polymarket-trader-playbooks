```python
import numpy as np
import pandas as pd

# Identify the 17 topic-allocation columns
topic_cols = [col for col in df.columns if col.startswith("topic_")]

# --- Topic-specialization features ---
df["top_topic_share"] = df[topic_cols].max(axis=1)

# Herfindahl-Hirschman Index:
# 1.00 = all activity in one topic; lower values = more diversified
df["topic_hhi"] = (df[topic_cols] ** 2).sum(axis=1)

# Effective number of topics:
# 1 = pure specialist; larger values = more diversified
df["effective_topics"] = 1 / df["topic_hhi"]

# Normalized entropy:
# 0 = pure specialist; 1 = maximally diversified across all 17 topics
topic_weights = df[topic_cols].replace(0, np.nan)
df["topic_entropy"] = -(
    topic_weights * np.log(topic_weights)
).sum(axis=1) / np.log(len(topic_cols))

# Trader's largest topic category
df["dominant_topic"] = (
    df[topic_cols]
    .idxmax(axis=1)
    .str.replace("topic_", "", regex=False)
)

# Secondary label indicator; do NOT use this in clustering later
df["is_sharp"] = (df["trader_label"] == "sharp").astype(int)

# Log versions for heavily skewed behavior variables
for col in [
    "trader_volume",
    "transaction_count",
    "transactions_per_day",
    "volume_per_day",
    "markets_per_day",
    "mean_tx_value"
]:
    df[f"log_{col}"] = np.log1p(df[col])

# Confirm the new features look reasonable
feature_summary = (
    df[[
        "top_topic_share",
        "topic_hhi",
        "effective_topics",
        "topic_entropy",
        "transactions_per_day",
        "markets_per_day",
        "mean_tx_value",
        "trader_ppv"
    ]]
    .describe(percentiles=[0.25, 0.5, 0.75])
    .T
    [["min", "25%", "50%", "75%", "max"]]
    .round(3)
)

display(feature_summary)

print("Most common dominant topics:")
display(
    df["dominant_topic"]
    .value_counts()
    .head(10)
    .rename_axis("Dominant topic")
    .reset_index(name="Number of traders")
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Create understandable specialization groups based on each trader's largest topic share
df["specialization_group"] = np.select(
    [
        df["top_topic_share"] >= 0.999999,
        df["top_topic_share"] >= 0.75,
        df["top_topic_share"] >= 0.50,
        df["top_topic_share"] >= 0.25
    ],
    [
        "Pure specialist\n(100% one topic)",
        "Highly concentrated\n(75% to <100%)",
        "Moderately concentrated\n(50% to <75%)",
        "Diversified\n(25% to <50%)"
    ],
    default="Very diversified\n(<25%)"
)

group_order = [
    "Pure specialist\n(100% one topic)",
    "Highly concentrated\n(75% to <100%)",
    "Moderately concentrated\n(50% to <75%)",
    "Diversified\n(25% to <50%)",
    "Very diversified\n(<25%)"
]

# Compare performance and sharp-trader share across specialization groups
specialization_results = (
    df.groupby("specialization_group", observed=False)
    .agg(
        traders=("trader", "size"),
        median_ppv=("trader_ppv", "median"),
        sharp_rate=("is_sharp", "mean"),
        median_volume=("trader_volume", "median"),
        median_effective_topics=("effective_topics", "median")
    )
    .reindex(group_order)
    .reset_index()
)

specialization_results["median_ppv_percent"] = (
    specialization_results["median_ppv"] * 100
)

specialization_results["sharp_rate_percent"] = (
    specialization_results["sharp_rate"] * 100
)

display(
    specialization_results[[
        "specialization_group",
        "traders",
        "median_effective_topics",
        "median_volume",
        "median_ppv_percent",
        "sharp_rate_percent"
    ]].round(3)
)

# Create the first two results figures
sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(
    data=specialization_results,
    x="specialization_group",
    y="median_ppv_percent",
    ax=axes[0],
    color="#B22222"
)
axes[0].axhline(0, color="black", linewidth=1)
axes[0].set_title("Profit per Dollar Traded by Topic Specialization")
axes[0].set_xlabel("")
axes[0].set_ylabel("Median profit per dollar traded (%)")
axes[0].tick_params(axis="x", rotation=15)

sns.barplot(
    data=specialization_results,
    x="specialization_group",
    y="sharp_rate_percent",
    ax=axes[1],
    color="#1F4E79"
)
axes[1].axhline(16.5, color="black", linestyle="--", linewidth=1)
axes[1].text(0.05, 17.2, "Overall sharp-trader rate: 16.5%", fontsize=10)
axes[1].set_title("Sharp-Trader Share by Topic Specialization")
axes[1].set_xlabel("")
axes[1].set_ylabel("Traders labeled sharp (%)")
axes[1].tick_params(axis="x", rotation=15)

plt.tight_layout()
plt.show()
```

```python
import statsmodels.api as sm
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Shorter group names for modeling
spec_map = {
    "Pure specialist\n(100% one topic)": "Pure specialist",
    "Highly concentrated\n(75% to <100%)": "Highly concentrated",
    "Moderately concentrated\n(50% to <75%)": "Moderately concentrated",
    "Diversified\n(25% to <50%)": "Diversified",
    "Very diversified\n(<25%)": "Very diversified"
}

df["spec_model_group"] = df["specialization_group"].map(spec_map)

# Create control variables
df["log_volume"] = np.log1p(df["trader_volume"])
df["log_transactions"] = np.log1p(df["transaction_count"])
df["log_transactions_day"] = np.log1p(df["transactions_per_day"])
df["log_markets_day"] = np.log1p(df["markets_per_day"])

controls = [
    "log_volume",
    "log_transactions",
    "log_transactions_day",
    "log_markets_day"
]

# Use pure specialists as the reference group
comparison_groups = [
    "Highly concentrated",
    "Moderately concentrated",
    "Diversified",
    "Very diversified"
]

model_data = df[["is_sharp", "spec_model_group"] + controls].dropna().copy()

for group in comparison_groups:
    model_data[f"spec_{group}"] = (
        model_data["spec_model_group"] == group
    ).astype(int)

dummy_cols = [f"spec_{group}" for group in comparison_groups]

X = model_data[controls + dummy_cols].astype(float)
X = sm.add_constant(X, has_constant="add")
y = model_data["is_sharp"]

# Logistic regression
sharp_model = sm.GLM(
    y,
    X,
    family=sm.families.Binomial()
).fit()

# Display specialization coefficients as odds ratios
coefficient_table = pd.DataFrame({
    "Group vs. pure specialist": comparison_groups,
    "Odds ratio": np.exp(sharp_model.params[dummy_cols].values),
    "P-value": sharp_model.pvalues[dummy_cols].values
})

display(coefficient_table.round({
    "Odds ratio": 3,
    "P-value": 4
}))

# Calculate adjusted sharp probabilities with controls held at their medians
control_medians = model_data[controls].median()

all_groups = ["Pure specialist"] + comparison_groups
adjusted_results = []

for group in all_groups:
    row = pd.DataFrame(0.0, index=[0], columns=X.columns)
    row["const"] = 1.0

    for control in controls:
        row[control] = control_medians[control]

    if group != "Pure specialist":
        row[f"spec_{group}"] = 1.0

    adjusted_probability = sharp_model.predict(row)[0] * 100

    adjusted_results.append({
        "Specialization group": group,
        "Adjusted sharp probability": adjusted_probability
    })

adjusted_results = pd.DataFrame(adjusted_results)
display(adjusted_results.round(2))

# Plot adjusted probabilities
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=adjusted_results,
    x="Specialization group",
    y="Adjusted sharp probability",
    color="#31597D"
)

plt.axhline(
    df["is_sharp"].mean() * 100,
    color="black",
    linestyle="--",
    linewidth=1,
    label="Overall sharp rate"
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f%%", padding=3)

plt.title("Adjusted Probability of Being a Sharp Trader")
plt.xlabel("Topic-specialization group")
plt.ylabel("Adjusted sharp probability (%)")
plt.xticks(rotation=20, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import RobustScaler
from sklearn.cluster import MiniBatchKMeans
from sklearn.metrics import silhouette_score
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Behavior-only clustering features
# P&L, trader_ppv, is_sharp, and trader_label are intentionally excluded
cluster_features = pd.DataFrame({
    "Trading frequency":
        np.log1p(df["transactions_per_day"]),

    "Daily volume":
        np.log1p(df["volume_per_day"]),

    "Market breadth":
        np.log1p(df["markets_per_day"]),

    "Average trade size":
        np.log1p(df["mean_tx_value"]),

    "Trading experience":
        np.log1p(df["transaction_count"]),

    "Execution depth":
        np.log1p(df["price_levels_per_transaction"]),

    "Price delta":
        df["mean_delta"],

    "Topic diversification":
        df["topic_entropy"]
}, index=df.index)

print("Missing values before cleaning:")
print(cluster_features.isna().sum())

# Replace infinite values and impute any remaining missing values with medians
cluster_features = cluster_features.replace([np.inf, -np.inf], np.nan)
cluster_features = cluster_features.fillna(cluster_features.median())

# Robust scaling reduces the influence of extreme traders
scaler = RobustScaler()
X_scaled = scaler.fit_transform(cluster_features)

print("\nClustering matrix shape:", X_scaled.shape)

# Use a representative sample for cluster selection
rng = np.random.default_rng(42)
selection_size = min(100000, len(df))
selection_idx = rng.choice(
    len(df),
    size=selection_size,
    replace=False
)

X_selection = X_scaled[selection_idx]

# Smaller sample for silhouette calculations
silhouette_size = min(5000, selection_size)
silhouette_idx = rng.choice(
    selection_size,
    size=silhouette_size,
    replace=False
)

results = []

for k in range(2, 9):
    model = MiniBatchKMeans(
        n_clusters=k,
        random_state=42,
        batch_size=4096,
        n_init=10
    )

    labels = model.fit_predict(X_selection)

    inertia = model.inertia_

    silhouette = silhouette_score(
        X_selection[silhouette_idx],
        labels[silhouette_idx]
    )

    results.append({
        "Number of clusters": k,
        "Inertia": inertia,
        "Silhouette score": silhouette
    })

cluster_selection = pd.DataFrame(results)
display(cluster_selection.round(4))

# Plot cluster-selection metrics
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(
    cluster_selection["Number of clusters"],
    cluster_selection["Inertia"],
    marker="o",
    color="#A63232"
)
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("Number of clusters")
axes[0].set_ylabel("Within-cluster inertia")

axes[1].plot(
    cluster_selection["Number of clusters"],
    cluster_selection["Silhouette score"],
    marker="o",
    color="#31597D"
)
axes[1].set_title("Cluster Separation")
axes[1].set_xlabel("Number of clusters")
axes[1].set_ylabel("Silhouette score")

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import RobustScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Cap extreme behavioral observations so clusters are not defined only by outliers
lower_bounds = cluster_features.quantile(0.005)
upper_bounds = cluster_features.quantile(0.995)

cluster_features_capped = cluster_features.clip(
    lower=lower_bounds,
    upper=upper_bounds,
    axis=1
)

# Scale features robustly
scaler = RobustScaler()
X_capped_scaled = scaler.fit_transform(cluster_features_capped)

# Use a 75,000-trader sample for exact K-means selection
rng = np.random.default_rng(42)

selection_size = 75000
selection_idx = rng.choice(
    len(df),
    size=selection_size,
    replace=False
)

X_selection = X_capped_scaled[selection_idx]

results = []
cluster_size_records = []

for k in range(2, 8):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20,
        algorithm="lloyd"
    )

    labels = model.fit_predict(X_selection)

    # Use 5,000 observations for silhouette calculation
    silhouette = silhouette_score(
        X_selection,
        labels,
        sample_size=5000,
        random_state=42
    )

    counts = pd.Series(labels).value_counts(normalize=True)

    results.append({
        "Number of clusters": k,
        "Inertia": model.inertia_,
        "Silhouette score": silhouette,
        "Smallest cluster share": counts.min() * 100,
        "Largest cluster share": counts.max() * 100
    })

    for cluster_id, share in counts.items():
        cluster_size_records.append({
            "Number of clusters": k,
            "Cluster": cluster_id,
            "Share of traders (%)": share * 100
        })

cluster_selection_exact = pd.DataFrame(results)
cluster_sizes = pd.DataFrame(cluster_size_records)

display(cluster_selection_exact.round(3))

print("Cluster shares for the 2-, 3-, and 4-cluster solutions:")
display(
    cluster_sizes[
        cluster_sizes["Number of clusters"].isin([2, 3, 4])
    ]
    .sort_values(["Number of clusters", "Cluster"])
    .round(2)
)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].plot(
    cluster_selection_exact["Number of clusters"],
    cluster_selection_exact["Inertia"],
    marker="o",
    color="#A63232"
)
axes[0].set_title("Exact K-Means Elbow Method")
axes[0].set_xlabel("Number of clusters")
axes[0].set_ylabel("Within-cluster inertia")

axes[1].plot(
    cluster_selection_exact["Number of clusters"],
    cluster_selection_exact["Silhouette score"],
    marker="o",
    color="#31597D"
)
axes[1].set_title("Cluster Separation")
axes[1].set_xlabel("Number of clusters")
axes[1].set_ylabel("Silhouette score")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# BUILD FIVE BALANCED PLAYBOOK DIMENSIONS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

# Percentile-rank transformation:
# 0 = low relative to other traders
# 1 = high relative to other traders
def percentile_rank(series):
    clean = (
        series.replace([np.inf, -np.inf], np.nan)
              .fillna(series.median())
    )
    return clean.rank(method="average", pct=True)

playbook_dimensions = pd.DataFrame(index=df.index)

# 1. Tempo: overall intensity and frequency
playbook_dimensions["tempo"] = (
    percentile_rank(df["transactions_per_day"]) +
    percentile_rank(df["volume_per_day"]) +
    percentile_rank(df["transaction_count"])
) / 3

# 2. Market breadth: number of markets entered
playbook_dimensions["market_breadth"] = percentile_rank(
    df["markets_per_day"]
)

# 3. Conviction and sizing: average capital per transaction
playbook_dimensions["conviction_sizing"] = percentile_rank(
    df["mean_tx_value"]
)

# 4. Execution aggressiveness
playbook_dimensions["execution_aggressiveness"] = (
    percentile_rank(df["price_levels_per_transaction"]) +
    percentile_rank(df["mean_delta"])
) / 2

# 5. Topic diversification
playbook_dimensions["topic_diversification"] = percentile_rank(
    df["topic_entropy"]
)

print("Playbook-dimension summary:")
display(playbook_dimensions.describe().round(3))
```

```python
# ============================================================
# RE-EVALUATE THE NUMBER OF PLAYBOOKS
# ============================================================

# Standardize the five conceptual dimensions
scaler = StandardScaler()
X_playbook = scaler.fit_transform(playbook_dimensions)

# Use a reproducible sample for model selection
selection_sample = playbook_dimensions.sample(
    n=100_000,
    random_state=42
).index

X_selection = X_playbook[selection_sample]

cluster_results = []
cluster_share_tables = {}

for k in range(2, 8):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20,
        max_iter=500
    )

    labels = model.fit_predict(X_selection)

    # Calculate silhouette on a smaller sample for speed
    silhouette = silhouette_score(
        X_selection,
        labels,
        sample_size=10_000,
        random_state=42
    )

    shares = (
        pd.Series(labels)
        .value_counts(normalize=True)
        .mul(100)
        .sort_values(ascending=False)
    )

    cluster_results.append({
        "Number of clusters": k,
        "Inertia": model.inertia_,
        "Silhouette score": silhouette,
        "Smallest cluster share": shares.min(),
        "Largest cluster share": shares.max()
    })

    cluster_share_tables[k] = shares

cluster_results = pd.DataFrame(cluster_results)

display(
    cluster_results.style.format({
        "Inertia": "{:,.0f}",
        "Silhouette score": "{:.3f}",
        "Smallest cluster share": "{:.2f}%",
        "Largest cluster share": "{:.2f}%"
    })
)

# Plot diagnostics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(
    cluster_results["Number of clusters"],
    cluster_results["Inertia"],
    marker="o",
    color="#b23a37"
)
axes[0].set_title("Playbook Clustering: Elbow Method")
axes[0].set_xlabel("Number of clusters")
axes[0].set_ylabel("Within-cluster inertia")
axes[0].grid(alpha=0.3)

axes[1].plot(
    cluster_results["Number of clusters"],
    cluster_results["Silhouette score"],
    marker="o",
    color="#315b7d"
)
axes[1].set_title("Playbook Cluster Separation")
axes[1].set_xlabel("Number of clusters")
axes[1].set_ylabel("Silhouette score")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Display shares for plausible solutions
for k in [3, 4, 5, 6]:
    print(f"\n{k}-cluster solution:")
    display(
        cluster_share_tables[k]
        .rename("Share of traders (%)")
        .to_frame()
        .round(2)
    )
```

```python
# ============================================================
# FINAL FOUR-PLAYBOOK CLUSTERING MODEL
# ============================================================

from sklearn.cluster import KMeans

FINAL_K = 4

final_model = KMeans(
    n_clusters=FINAL_K,
    random_state=42,
    n_init=20,
    max_iter=500
)

df["playbook_cluster"] = final_model.fit_predict(X_playbook)

print("Final cluster counts:")
display(
    df["playbook_cluster"]
    .value_counts()
    .sort_index()
    .rename("Number of traders")
    .to_frame()
)

print("\nFinal cluster shares:")
display(
    df["playbook_cluster"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .rename("Share of traders (%)")
    .round(2)
    .to_frame()
)
```

```python
# ============================================================
# FINAL STATISTICAL VALIDATION
# ============================================================

import numpy as np
import pandas as pd

from scipy.stats import (
    kruskal,
    chi2_contingency,
    binom
)

# ------------------------------------------------------------
# 1. Kruskal-Wallis test: Do PPV distributions differ?
# ------------------------------------------------------------

ppv_groups = [
    df.loc[
        df["playbook_cluster"] == cluster,
        "trader_ppv"
    ].dropna().values
    for cluster in sorted(df["playbook_cluster"].unique())
]

kw_stat, kw_p = kruskal(*ppv_groups)

n_total = sum(len(group) for group in ppv_groups)
k_groups = len(ppv_groups)

# Epsilon-squared effect size
kw_epsilon_squared = (
    (kw_stat - k_groups + 1) /
    (n_total - k_groups)
)

# ------------------------------------------------------------
# 2. Chi-square test: Does label composition differ?
# ------------------------------------------------------------

label_order = ["awful", "bad", "good", "sharp"]

contingency_table = pd.crosstab(
    df["playbook_cluster"],
    df["trader_label"]
).reindex(columns=label_order)

chi2_stat, chi2_p, chi2_dof, expected = chi2_contingency(
    contingency_table
)

# Cramer's V effect size
rows, columns = contingency_table.shape

cramers_v = np.sqrt(
    chi2_stat /
    (
        contingency_table.to_numpy().sum() *
        min(rows - 1, columns - 1)
    )
)

test_results = pd.DataFrame({
    "Test": [
        "Kruskal-Wallis: PPV distributions",
        "Chi-square: trader-label composition"
    ],
    "Statistic": [
        kw_stat,
        chi2_stat
    ],
    "Degrees of freedom": [
        k_groups - 1,
        chi2_dof
    ],
    "P-value": [
        kw_p,
        chi2_p
    ],
    "Effect size": [
        kw_epsilon_squared,
        cramers_v
    ],
    "Effect-size measure": [
        "Epsilon squared",
        "Cramer's V"
    ]
})

display(
    test_results.style.format({
        "Statistic": "{:,.2f}",
        "P-value": "{:.3e}",
        "Effect size": "{:.3f}"
    })
)
```

```python
# ============================================================
# 95% CONFIDENCE INTERVALS FOR MEDIAN PPV
# ============================================================

median_ci_rows = []

for cluster, group in df.groupby("playbook_cluster"):
    values = np.sort(
        group["trader_ppv"]
        .dropna()
        .to_numpy()
    )

    n = len(values)

    lower_index = int(
        binom.ppf(0.025, n, 0.5)
    )

    upper_index = int(
        binom.ppf(0.975, n, 0.5)
    )

    # Keep indices within the array
    lower_index = max(0, min(lower_index, n - 1))
    upper_index = max(0, min(upper_index, n - 1))

    median_ci_rows.append({
        "Playbook": playbook_names[cluster],
        "Traders": n,
        "Median PPV (bps)": np.median(values) * 10_000,
        "95% CI lower (bps)": values[lower_index] * 10_000,
        "95% CI upper (bps)": values[upper_index] * 10_000
    })

median_ci = pd.DataFrame(median_ci_rows)

display(
    median_ci.style.format({
        "Median PPV (bps)": "{:.2f}",
        "95% CI lower (bps)": "{:.2f}",
        "95% CI upper (bps)": "{:.2f}"
    })
)
```
